```python
import pandas as pd

df = pd.DataFrame({
    'Marks': [45, 50, 52, 48, 55, 51, 49, 95, 47, 53]
})
```

```python
# calculate Q1 and Q3

Q1 = df['Marks'].quantile(0.25)
Q3 = df['Marks'].quantile(0.75)
```

```python
# calculate IQR

IQR = Q3 - Q1
```

```python
# calculate limits

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
print(lower)
print(upper)
```

```python
# identify outliers

outliers = df[
    (df['Marks']<lower) |
    (df['Marks']>upper)
]

print("Outliers: ")
print(outliers)
```

```python
# handling outliers using capping 

df['Marks'] = df['Marks'].clip(
    lower=lower,
    upper=upper
)

print("Data after handling outliers: ")
print(df)
```
