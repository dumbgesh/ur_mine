```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# our dataset

data = {
    "student_hours" : [1, 2, 3, 4, 5],
    "marks" : [35, 40, 50, 55, 70]
}

df = pd.DataFrame(data)
df
```

```python
x = df["student_hours"].to_numpy()
y = df["marks"].to_numpy()

x_mean = np.mean(x)
y_mean = np.mean(y)

print("Mean Study Hours: ", x_mean)
print("Mean Marks: ", y_mean)
```

```python
# calculate deviations from the mean

x_deviation = x - x_mean
y_deviation = y - y_mean

# calculate the slope (m)

numerator = np.sum(x_deviation * y_deviation)
denominator = np.sum(x_deviation ** 2)

m = numerator / denominator

print("Slope (m): ", m)
```

```python
# calculate the intercept (c)

c = y_mean - (m * x_mean)
print("Intercept (c): ", c)
```

```python
# we'll make predictions

y_pred = m * x + c
print("Predicted Marks: ", y_pred)
```

```python
df["predicted_marks"] = y_pred
df["error"] = y - y_pred

df
```

```python
# predict for a new student

study_hours = 6
predicted_marks = m * study_hours + c
print(f"Predicted marks for a student who studies {study_hours} hours: {predicted_marks}")
```

```python
# visualization

plt.scatter(x, y, label="Actual marks")

plt.plot(x, y_pred, label="Regression line")

plt.xlabel("Study hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# evaluation

errors = y - y_pred

mae = np.mean(np.abs(errors))
mse = np.mean(errors ** 2)

print("MAE:", mae)
print("MSE:", mse)
```
